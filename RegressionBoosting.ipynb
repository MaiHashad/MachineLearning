```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

#Loading the dataset
dataset = pd.read_csv('/content/drive/MyDrive/SEIS763ML/DataML/salary.csv')

#Extracting the independent and dependent variables
X = dataset.iloc[:,1:2]
y = dataset.iloc[:,-1]

#Converting to numpy array
X = np.array(X)
y = np.array(y)
```

```python
#Fitting Classifier to Training Set
from sklearn.ensemble import AdaBoostRegressor 
regObjAda = AdaBoostRegressor()
regObjAda.fit(X, y)
```

```python
#Fitting Classifier to Training Set
from sklearn.ensemble import GradientBoostingRegressor 
regObjGB = GradientBoostingRegressor()
regObjGB.fit(X, y)
```

```python
from xgboost import XGBRegressor
regObjXGB = XGBRegressor()
regObjXGB.fit(X, y)
```

```python
#Visualizing the Decision Tree Regression in higher resolution
X_grid = np.arange(min(X), max(X), 0.01)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y, color = 'red')
plt.plot(X_grid, regObjAda.predict(X_grid), color = 'blue')
plt.title('Results')
plt.xlabel('Level')
plt.ylabel('Salary')
plt.show()


#Predicting with the Model 
```
